```python
#Importing the basic library
import numpy as np
from matplotlib import pyplot as plt
```

```python
x=np.arange(1,11)
```

```python
x
```

```python
y=2*x
```

```python
y
```

```python
#Line Chart
plt.plot(x,y)
plt.title("Line Chart")
plt.xlabel("x-axis")

plt.ylabel("y-axis")
plt.show
```

```python
#BAR CHART
plt.bar(x,y)
plt.title("BarChart")
plt.xlabel("x-axis")

plt.ylabel("y-axis")
plt.show
```

```python
#Scatter plot
plt.scatter(x,y)
plt.title("scatter plot")
plt.xlabel("x-axis")

plt.ylabel("y-axis")
plt.show
```
